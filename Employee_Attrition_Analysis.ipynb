TASK-3 Data Visualization

### Task 3 – Data Visualization
Visual representation of important employee attrition patterns.

#### Task 3.1 – Attrition Distribution Chart

```python
import pandas as pd
import matplotlib.pyplot as plt
df["Attrition"].value_counts().plot(kind="bar")
plt.title("Attrition Distribution")
plt.xlabel("Attrition(0)=no,(1)=Yes")
plt.ylabel("no. of Employees")
plt.show()
```

📊 Chart 2 – Department Distribution


#### Task 3.2 – Department Distribution Chart

```python
df["Department"].value_counts().plot(kind="bar")
plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("No. of Employees")
plt.xticks(rotation=45)
plt.show()
```

Chart 3 – Gender Distribution

#### Task 3.3 – Gender Distribution Chart

```python
df["Gender"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```

Chart 4 – OverTime vs Attrition

Markdown:

#### Task 3.4 – OverTime vs Attrition Chart

```python
(pd.crosstab(df["OverTime"], df["Attrition"], normalize="index")*100)[1].plot(kind="bar")
plt.title("Attrition Rate by OverTime")
plt.xlabel("OverTime")
plt.ylabel("Attrition %")
plt.show()
```

Chart 5 – Job Role Distribution

#### Task 3.5 – Job Role Distribution Chart
